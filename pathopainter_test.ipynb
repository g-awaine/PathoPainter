```python
"""
PathoPainter single-instance sampling script.

Drop this file next to the original sampling.py (same directory, so the
`utils`, `ldm.models...`, `ldm.util` imports resolve). Fill in the four
paths below, then run: python sampling_single.py
"""
import torch
import numpy as np
from pathlib import Path
from PIL import Image
from utils import get_model
from ldm.models.diffusion.plms import PLMSSampler

# ---------------------------------------------------------------------
# 1. Point these at ONE real example
# ---------------------------------------------------------------------
image_path = "path/to/image/patch_0001.png"          # H&E tile (RGB)
mask_path = "path/to/mask/patch_0001.png"             # binary mask, 0/255
feat_path = "path/to/ssl_features/patch_0001.npy"     # HIPT region embedding
output_path = "output.png"

# To use the "regional embedding sampling" trick from the paper (pull the
# foreground embedding from a DIFFERENT image in the same lesion cluster,
# rather than reusing this image's own embedding), just point feat_path at
# that other image's .npy file instead.

# ---------------------------------------------------------------------
# 2. Sampling parameters (same defaults as the original script)
# ---------------------------------------------------------------------
scale = 2
ddim_steps = 50
shape = [3, 64, 64]
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")

# ---------------------------------------------------------------------
# 3. Load model
# ---------------------------------------------------------------------
model_path = Path("logs/08-07T14-37_pathopainter/")
model, config = get_model(model_path, device, "last.ckpt")
sampler = PLMSSampler(model)

# ---------------------------------------------------------------------
# 4. Load and preprocess the single image / mask / feature
#    (mirrors TCGADataset.__getitem__ exactly, just without the DataLoader)
# ---------------------------------------------------------------------
img = np.array(Image.open(image_path).convert("RGB"))
image_np = (img / 127.5 - 1.0).astype(np.float32)                  # [-1, 1], HWC
image = torch.from_numpy(image_np).permute(2, 0, 1).unsqueeze(0)   # 1x3xHxW

mask_np = (np.array(Image.open(mask_path)) / 255.0).astype(np.float32)
mask = torch.from_numpy(mask_np).unsqueeze(0).unsqueeze(0)         # 1x1xHxW

feat_patch = torch.from_numpy(np.load(feat_path)).unsqueeze(0)     # 1 x feat_dim

image = image.to(device)
mask = mask.to(device)
feat_patch = feat_patch.to(device)
batch_size = 1

# ---------------------------------------------------------------------
# 5. Inference
# ---------------------------------------------------------------------
with torch.no_grad(), model.ema_scope():
    # Background-only image (region under the mask removed)
    masked_img = (image * (1 - mask)).to(torch.float32)

    # Encode background to latent
    encoder_posterior_mask = model.encode_first_stage(masked_img)
    z_masked = model.get_first_stage_encoding(encoder_posterior_mask).detach()

    # Resize mask to latent resolution, build cond_mask
    cc_mask = torch.nn.functional.interpolate(mask, size=z_masked.shape[-2:])
    cond_mask = torch.cat((z_masked, cc_mask), dim=1)  # 1 x 4 x 64 x 64

    batch = {
        "image": torch.tensor((127.5 * (image + 1)), dtype=torch.uint8).to(device),
        "mask": mask.to(torch.uint8),
        "feat_patch": feat_patch,
        "cond_mask": cond_mask,
    }

    batch_uncond = {**batch}
    batch_uncond["feat_patch"] = torch.zeros_like(batch["feat_patch"])

    cc = model.get_learned_conditioning(batch)
    uc = model.get_learned_conditioning(batch_uncond)

    samples_ddim, _ = sampler.sample(
        S=ddim_steps,
        conditioning=cc,
        batch_size=batch_size,
        shape=shape,
        verbose=False,
        unconditional_guidance_scale=scale,
        unconditional_conditioning=uc,
        cond_mask=batch["cond_mask"],
    )

    x_samples_ddim = model.decode_first_stage(samples_ddim)
    x_samples_ddim = torch.clamp((x_samples_ddim + 1.0) / 2.0, min=0.0, max=1.0)
    x_samples_ddim = (x_samples_ddim * 255).to(torch.uint8).cpu()
    x_samples_ddim = x_samples_ddim.permute(0, 2, 3, 1).numpy()

    out_path = Path(output_path)
    out_path.parent.mkdir(parents=True, exist_ok=True)
    Image.fromarray(x_samples_ddim[0]).save(out_path)

print(f"Saved synthetic image to {output_path}")
```
